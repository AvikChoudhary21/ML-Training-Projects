```python
#import required libraries and load dataset
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("/content/zomato_eda_mini_project.csv")
df.head()
```

```python
#Data Inspection
df.shape()
```

```python
#info 
df.info()
```

```python
#Checking Missing values
df.isnull().sum()
```

```python
#Checking duplicates
df.duplicated().sum()
```

```python
df["Cuisines"] =  df["Cuisines"].fillna("Unknown")
df["Average Cost for two"] = df["Average Cost for two"].fillna(df["Average Cost for two"].median)
df["Rating"] = df["Aggregate rating"].astype(float)
```

```python
df["Rating"].isnull().sum()
```

```python
#Visualization
#Chart 1 - Countplot - Which price range is most common

plt.figure(figsize=(9, 5))
sns.countplot(data=df, x = "Price Range")
plt.title("Number of Restaurants by price range")
plt.xlabel("Price Range")
plt.ylabel("Number of Restaurants")
plt.show()
```

```python
#Chart 2 - Which price range has better ratings
#Bar Plot

plt.figure(figsize=(8,7))
sns.barplot(data=df, x = "Price Range", y = "Rating")
plt.title("Restaurants rating by price range")
plt.xlabel("Price Range")
plt.ylabel("Average Ratings")
plt.show()
```
